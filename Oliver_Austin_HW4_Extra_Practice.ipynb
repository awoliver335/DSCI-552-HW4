## 1. ISLR 4.8.4 - Curse of dimensionality

Since the observations are uniform over the unit cube, the expected fraction
inside a region is its volume. I assume the neighborhood fits inside the
feature range, as in the exercise.

**(a)** With one feature, an interval of length $0.1$ contains about
$0.1=$ **10%** of the observations.

**(b)** With two features, the square has area $0.1^2=0.01$, so it contains
about **1%** of the observations.

**(c)** With 100 independent uniform features, the volume is
$0.1^{100}=10^{-100}$. This is effectively zero for a practical dataset.

**(d)** In $p$ dimensions, this neighborhood contains $n(0.1)^p$ observations
on average. That number drops very quickly as $p$ increases. KNN then has to
use a larger neighborhood to find enough neighbors, meaning the neighbors
are no longer very close. This is the curse of dimensionality.

**(e)** If the side length is $a$, the volume is $a^p$. To include 10% of the
observations on average,

$$a^p=0.1 \quad\Rightarrow\quad a=0.1^{1/p}.$$

| Dimensions $p$ | Side length $a$ |
|---:|---:|
| 1 | 0.1000 |
| 2 | 0.3162 |
| 100 | 0.9772 |

At 100 dimensions, each side covers about 97.72% of the feature's range.
That is a very large neighborhood, which shows why finding local neighbors
is difficult in high dimensions. These volume calculations assume the cube
stays inside $[0,1]^p$; a cube extending outside it would contain less data.

## 2. ISLR 4.8.9 - Odds and probabilities

**(a)** Starting with $o=p/(1-p)$ and solving for $p$ gives
$p=o/(1+o)$. When the odds are 0.37,

$$p=\frac{0.37}{1+0.37}=0.270073\approx\boxed{27.01\%}.$$

**(b)** When the probability is 0.16, the odds are

$$o=\frac{0.16}{1-0.16}=\frac{0.16}{0.84}=\frac4{21}
\approx\boxed{0.190476}.$$

So the odds of default are about **0.1905 to 1**, or **4 to 21**.
